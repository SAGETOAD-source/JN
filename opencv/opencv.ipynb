```python
import cv2 
print("Welcome to the world of Computer Vision where we make colorful images to grayscale!")
input_image = input("Please give the relative path of your image...\n")
image = cv2.imread(input_image)
gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
print("Do you want to see the image of save the image?")
press = input("Write 'see' to see the image and Write 'save' to save it\n")
if press=='save':
    output_name = input("What name do you want to save your image as?\n")  
    cv2.imwrite(output_name,gray)
    print('Image saved successfully')
elif press =='see':
    cv2.imshow("Grayscaled",gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2

image = cv2.imread("images/krishu.jpeg")

if image is None:
    print("image not found")
else:
    flipped_horizontal = cv2.flip(image, 1)
    flipped_vertical = cv2.flip(image, 0)
    flipped_both = cv2.flip(image, -1)

    cv2.namedWindow("Original Image", cv2.WINDOW_NORMAL)
    cv2.namedWindow("Flipped Horizontally", cv2.WINDOW_NORMAL)
    cv2.namedWindow("Flipped Vertically", cv2.WINDOW_NORMAL)
    cv2.namedWindow("Flipped Both", cv2.WINDOW_NORMAL)

    # Set window size
    cv2.resizeWindow("Original Image", 800, 600)
    cv2.resizeWindow("Flipped Horizontally", 800, 600)
    cv2.resizeWindow("Flipped Vertically", 800, 600)
    cv2.resizeWindow("Flipped Both", 800, 600)

    cv2.imshow("Original Image", image)
    cv2.imshow("Flipped Horizontally", flipped_horizontal)
    cv2.imshow("Flipped Vertically", flipped_vertical)
    cv2.imshow("Flipped Both", flipped_both)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2
image = cv2.imread("images/krishu.jpeg")
if image is None:
    print("image not found")
else:
    print("image loaded")   
    cv2.resizeWindow("Image with Line", 800, 600)
    image = cv2.imread("images/krishu.jpeg")
    if image is None:
        print("image not found")
    else:
        print("image loaded")
        cv2.namedWindow("Image with Line", cv2.WINDOW_NORMAL)
        cv2.resizeWindow("Image with Line", 800, 600)
        pt1 = (50, 100)
        ptt2 = (300, 100)
        color = (0, 255, 0)
        thickness = 5

        cv2.line(image, pt1, ptt2, color, thickness)
        cv2.imshow("Image with Line", image)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    
```

```python
# pt1 and pt2
# pt1 is the top-left corner
# pt2 is the bottom-right corner

import cv2

image = cv2.imread("images/krishu.jpeg")

if image is None:
    print("image not found")
else:
    print("image loaded")

    cv2.namedWindow("Image with Rectangle", cv2.WINDOW_NORMAL)
    cv2.resizeWindow("Image with Rectangle", 800, 600)

    cv2.rectangle(image, (800, 450), (1130, 1000), (0, 255, 0), 3)
    cv2.putText(image, "This is a rectangle", (800, 450), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.imshow("Image with Rectangle", image)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# file input via user input like asking the file directory and file name and then reading the image from that path and then displaying it in a window and then saving it to a new file with a new name and then displaying the new image in a window and then closing all windows when the user presses any key.
file_input = input("Please give the relative path of your image...\n")
image = cv2.imread(file_input)
if image is None:
    print("Image not found. Please check the path and try again.")
else:
    cv2.imshow("Original Image", image)
    cv2.waitKey(0)

    output_name = input("What name do you want to save your image as?\n")
    cv2.imwrite(output_name, image)
    print('Image saved successfully')

    new_image = cv2.imread(output_name)
    cv2.imshow("New Image", new_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
