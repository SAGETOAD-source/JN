```python
import gc 
class myobject:
    def __init__(self,name):
        self.name = name
        print(f"object {self.name} created")
    def __del__(self):
        print(f"object {self.name} deleted")
obj1 = myobject("obj1")
obj2 = myobject("obj2")
obj1.ref = obj2 
obj2.ref = obj1

del obj1
del obj2

gc.collect()
```
