```python
class Backaccount:
    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
        else:
            print("Insufficient funds")
    def interest(self, day, amount):
        is_add = input("Do you want to add recent amount with previous amount? (yes/no): ").lower()
        if is_add == 'yes':
            self.balance += amount
            self.balance = self.balance + self.balance * 10.21 * day
            print("After expiry the total amount is", self.balance)
        else:
            self.balance2 = amount + amount * 10.21 * day
            print("After expiry the total amount is", self.balance2)

account = Backaccount("krishnendu", 5000)
print(account.name)
print(account.balance)
account.deposit(2000)
print(account.balance)
account.interest(100,510)
```

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def make_sound(self):
        print(f"{self.name} makes a sound: {self.sound}")

class Dog(Animal):
    def fetch(self):
        print(f"{self.name } is fetching the ball")

        
dog1 = Dog("Buddy", "Bark")
dog1.make_sound()


dog1.fetch()    
```

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):  # MAGIC METHOD
        return Money(self.amount + other.amount)

m1 = Money(100)
m2 = Money(200)
m3 = m1 + m2  # Automatically calls __add__

print(m3.amount)  # Output: 300
```

```python
class Box:
    def __init__(self, volume):
        self.volume = volume

    def __add__(self, other):  # Operator Overloading
        return Box(self.volume + other.volume)

    def __str__(self):
        return f"Box with volume {self.volume}"

b1 = Box(10)
b2 = Box(20)
b3 = b1 + b2  # Calls b1.__add__(b2)

print(b3)  # Output: Box with volume 30
```

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __add__(self, other):
        if isinstance(other, Point):
            return Point(self.x + other.x, self.y + other.y)
        return NotImplemented
    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)
print(p1 + p2)  # Output: (6, 8)
```

```python
class MyRange:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self   # returns the iterator object

    def __next__(self):
        if self.current >= self.end:
            raise StopIteration
        num = self.current
        self.current += 1
        return num
```

```python
class MyIterator:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 3:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration
```

```python
def square(n):
    for i in range(n):
        yield i ** 2
 
for i in square(3):
    print(i)
```

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for i in countdown(5):
    print(i)
 
```

```python
def add_all(*args):
    total = 0
    for num in args:
        total += num
    print(total)

add_all(1, 2, 3)         # 6
add_all(5, 10, 15, 20)   # 50
```

What is *args?
*args allows a function to accept any number of positional arguments — even if you don’t know how many there will be.

```python
def add_all(*args):
    total = 0
    for i in args:
        total += i
    print(total)

add_all(1, 2, 3)



```

```python
import threading
import time
def print_numbers():
    for i in range(5):
        time.sleep(2)
        print(f"Number:{i}")
def print_letter():
    for letter in "abcde":
        time.sleep(2)
        print(f"letter:{letter}")
##creating threads
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_letter)

t=time.time()
##starting  threads
t1.start()
t2.start()
##waiting for threads to complete
t1.join()
t2.join()
##print_numbers()
##print_letter()
finished_time=time.time()-t
print(f"Execution time without threading:{finished_time} seconds")                


```
